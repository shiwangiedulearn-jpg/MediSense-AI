```python
categorical = [
    "rbc",
    "pc",
    "pcc",
    "ba",
    "htn",
    "dm",
    "cad",
    "appet",
    "pe",
    "ane"
]

for col in categorical:
    print("=" * 40)
    print(col)
    print(df[col].unique())
```

```python
categorical = [
    "rbc",
    "pc",
    "pcc",
    "ba",
    "htn",
    "dm",
    "cad",
    "appet",
    "pe",
    "ane"
]

for col in categorical:
    df[col] = df[col].astype(str).str.strip()
    df[col] = df[col].replace("nan", pd.NA)
```

```python
numerical = [
    "age","bp","sg","al","su","bgr","bu",
    "sc","sod","pot","hemo","pcv","wc","rc"
]

for col in numerical:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df[col] = df[col].fillna(df[col].median())
```

```python
for col in categorical:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df["rbc"] = df["rbc"].map({
    "normal": 0,
    "abnormal": 1
})

df["pc"] = df["pc"].map({
    "normal": 0,
    "abnormal": 1
})

df["pcc"] = df["pcc"].map({
    "notpresent": 0,
    "present": 1
})

df["ba"] = df["ba"].map({
    "notpresent": 0,
    "present": 1
})

df["htn"] = df["htn"].map({
    "no": 0,
    "yes": 1
})

df["dm"] = df["dm"].map({
    "no": 0,
    "yes": 1
})

df["cad"] = df["cad"].map({
    "no": 0,
    "yes": 1
})

df["appet"] = df["appet"].map({
    "good": 0,
    "poor": 1
})

df["pe"] = df["pe"].map({
    "no": 0,
    "yes": 1
})

df["ane"] = df["ane"].map({
    "no": 0,
    "yes": 1
})
```
